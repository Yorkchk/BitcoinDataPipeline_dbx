```python
tables_in_schema = spark.catalog.listTables(dbName="bitcoinproject_workspace.default")
gold_tables = [table.name for table in tables_in_schema if "gold" in table.name]
```

```python
username = dbutils.secrets.get(scope="Connectors", key="username")
password = dbutils.secrets.get(scope="Connectors", key="sqlpassword")
jdbc_url = f"jdbc:sqlserver://sqlserver-bitcoin.database.windows.net:1433;database=bitcoin_data;user={username}@sqlserver-bitcoin;password={password};encrypt=true;trustServerCertificate=false;hostNameInCertificate=*.database.windows.net;loginTimeout=30;"
for table in gold_tables: 
  df = spark.read.table(table)

  (df.write
    .format("jdbc")
    .mode("overwrite")
    .option("url", jdbc_url)
    .option("dbtable", table)
    .option("user", username)
    .option("password", password)
    .option("driver", "com.microsoft.sqlserver.jdbc.SQLServerDriver")
    .save()
  )
```
